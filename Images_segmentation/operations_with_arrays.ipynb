В программе создан массив и записан в переменную boxes. Массив boxes имеет размер (100, 4), он содержит 100 ограничивающих рамок в формате (Xц, Yц, W, H).
Параметры рамок в относительных значениях.

Создайте функцию с названием xywh2xyxy. Данная функция принимает один обязательный параметр boxes, который содержит массив с ограничивающими рамками. Размер массива (num_boxes, xyhw), где num_boxes - количество ограничивающих рамок, а  xyhw - четыре параметра ограничивающих рамок в формате (Xц, Yц, W, H).

Функция xyhw2xyxy должна возвращать массив размером (num_boxes, xyxy), где xyxy - четыре параметра ограничивающих рамок в формате (Xлв, Yлв, Xпн, Yпн).

Новые параметры рамок в относительных значениях.

```python
def xywh2xyxy(boxes):
    new_boxes = np.full(boxes.shape, 0)
    new_boxes[:, 0] = boxes[:, 0] - boxes[:, 2] / 2
    new_boxes[:, 1] = boxes[:, 1] - boxes[:, 3] / 2
    new_boxes[:, 2] = boxes[:, 0] + boxes[:, 2] / 2
    new_boxes[:, 3] = boxes[:, 1] + boxes[:, 3] / 2
    
    return new_boxes


new_boxes = xywh2xyxy(boxes)
```

В программе создан массив и записан в переменную boxes. Массив boxes имеет размер (100, 4), он содержит 100 ограничивающих рамок в формате (Xлв, Yлв, Xпн, Yпн).
Параметры рамок в относительных значениях.

Создайте функцию с названием xyxy2xywh. Данная функция принимает один обязательный параметр boxes, который содержит массив с ограничивающими рамками. Размер массива (num_boxes, xyxy), где num_boxes - количество ограничивающих рамок, а  xyxy - четыре параметра ограничивающих рамок в формате (Xлв, Yлв, Xпн, Yпн).
Функция xyxy2xywh должна возвращать массив размером (num_boxes, xywh), где xywh - четыре параметра ограничивающих рамок в формате (Xц, Yц, W, H).

Новые параметры рамок в относительных значениях.

```python
def xyxy2xywh(boxes):
    new_boxes = np.zeros_like(boxes)
    new_boxes[:, 0] = boxes[:, 0] + (boxes[:, 2] - boxes[:, 0]) / 2
    new_boxes[:, 1] = boxes[:, 1] + (boxes[:, 3] - boxes[:, 1]) / 2
    new_boxes[:, 2] = boxes[:, 2] - boxes[:, 0]
    new_boxes[:, 3] = boxes[:, 3] - boxes[:, 1]

    return new_boxes


new_boxes = xyxy2xywh(boxes)
```
